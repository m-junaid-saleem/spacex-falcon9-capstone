# EDA with Data Visualization

Generates all exploratory-data-analysis charts (flight/payload vs site/orbit, success rate by orbit, yearly trend).

```python
import pandas as pd, matplotlib
matplotlib.use('Agg')
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches

df = pd.read_csv('spacex_launch_geo.csv')
df['Date'] = pd.to_datetime(df['Date'])
df['Year'] = df['Date'].dt.year
plt.rcParams['figure.dpi'] = 150
GREEN, RED = '#2e7d32', '#c62828'
colors = df['class'].map({1: GREEN, 0: RED})

# 1. Flight Number vs Launch Site
plt.figure(figsize=(8,5))
plt.scatter(df['Flight Number'], df['Launch Site'], c=colors, s=70, edgecolor='white')
plt.xlabel('Flight Number'); plt.ylabel('Launch Site')
plt.title('Flight Number vs. Launch Site')
plt.legend(handles=[mpatches.Patch(color=GREEN,label='Success'), mpatches.Patch(color=RED,label='Failure')])
plt.tight_layout(); plt.savefig('c1_flight_vs_site.png'); plt.close()

# 2. Payload vs Launch Site
plt.figure(figsize=(8,5))
plt.scatter(df['Payload Mass (kg)'], df['Launch Site'], c=colors, s=70, edgecolor='white')
plt.xlabel('Payload Mass (kg)'); plt.ylabel('Launch Site')
plt.title('Payload vs. Launch Site')
plt.legend(handles=[mpatches.Patch(color=GREEN,label='Success'), mpatches.Patch(color=RED,label='Failure')])
plt.tight_layout(); plt.savefig('c2_payload_vs_site.png'); plt.close()

# 3. Success rate by Orbit
plt.figure(figsize=(8,5))
orbit_rate = df.groupby('Orbit')['class'].mean().sort_values()
orbit_rate.plot(kind='bar', color='#1565c0')
plt.ylabel('Success Rate'); plt.xlabel('Orbit Type')
plt.title('Success Rate vs. Orbit Type')
plt.xticks(rotation=45, ha='right')
plt.tight_layout(); plt.savefig('c3_success_vs_orbit.png'); plt.close()

# 4. Flight Number vs Orbit
plt.figure(figsize=(8,5))
plt.scatter(df['Flight Number'], df['Orbit'], c=colors, s=70, edgecolor='white')
plt.xlabel('Flight Number'); plt.ylabel('Orbit Type')
plt.title('Flight Number vs. Orbit Type')
plt.legend(handles=[mpatches.Patch(color=GREEN,label='Success'), mpatches.Patch(color=RED,label='Failure')])
plt.tight_layout(); plt.savefig('c4_flight_vs_orbit.png'); plt.close()

# 5. Payload vs Orbit
plt.figure(figsize=(8,5))
plt.scatter(df['Payload Mass (kg)'], df['Orbit'], c=colors, s=70, edgecolor='white')
plt.xlabel('Payload Mass (kg)'); plt.ylabel('Orbit Type')
plt.title('Payload vs. Orbit Type')
plt.legend(handles=[mpatches.Patch(color=GREEN,label='Success'), mpatches.Patch(color=RED,label='Failure')])
plt.tight_layout(); plt.savefig('c5_payload_vs_orbit.png'); plt.close()

# 6. Yearly success trend
plt.figure(figsize=(8,5))
yearly = df.groupby('Year')['class'].mean()
plt.plot(yearly.index, yearly.values, marker='o', color='#1565c0', linewidth=2)
plt.xlabel('Year'); plt.ylabel('Average Success Rate')
plt.title('Launch Success Yearly Trend')
plt.grid(alpha=0.3)
plt.tight_layout(); plt.savefig('c6_yearly_trend.png'); plt.close()

print("EDA charts done")
print(df.groupby('Launch Site')['class'].agg(['count','sum']))
```
